```python
import json

with open(r"output\chunks.json", "r", encoding="utf-8") as f:
    chunks = json.load(f)


topics = sorted({
    chunk["topic"]
    for chunk in chunks
    if chunk.get("source_book") == "Ghurar al-Hikam"
})

print(f"Unique topics: {len(topics)}\n")

for topic in topics:
    print(topic)
```

```python
from google import genai

API_KEY = ''

client = genai.Client(api_key=API_KEY)

prompt = "Say hello in one short sentence."

print("\nFetching available models...\n")

working = []
failed = []

try:
    models = client.models.list()

    for model in models:
        # Only test models that support generateContent
        if "generateContent" not in (model.supported_actions or []):
            continue

        model_name = model.name

        # Remove "models/" prefix if present
        if model_name.startswith("models/"):
            model_name = model_name.replace("models/", "", 1)

        print("=" * 70)
        print(f"Testing: {model_name}")

        try:
            response = client.models.generate_content(
                model=model_name,
                contents=prompt
            )

            print("✅ WORKS")
            print("Response:", response.text)

            working.append(model_name)

        except Exception as e:
            print("❌ FAILED")
            print(type(e).__name__)
            print(str(e))

            failed.append(model_name)

except Exception as e:
    print("Could not retrieve model list.")
    print(type(e).__name__)
    print(e)


print("\n")
print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print("\n✅ WORKING MODELS:")
for model in working:
    print(f"  - {model}")

print("\n❌ FAILED MODELS:")
for model in failed:
    print(f"  - {model}")

print(f"\nWorking: {len(working)}")
print(f"Failed:  {len(failed)}")
```

```python
import json
import os

output_file = "ghurar_enriched.json"
progress_file = output_file + ".progress.json"

for path in [output_file, progress_file]:
    if os.path.exists(path):
        mtime = os.path.getmtime(path)
        import datetime
        print(f"{path}: last modified {datetime.datetime.fromtimestamp(mtime)}")
    else:
        print(f"{path}: DOES NOT EXIST")

enriched = json.load(open(output_file))
progress = json.load(open(progress_file))
print(f"\nEnriched file has {len(enriched)} chunks")
print(f"Progress file has {len(progress['done_keys'])} done_keys")
print(f"Match: {len(enriched) == len(progress['done_keys'])}")
```
